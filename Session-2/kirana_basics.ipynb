#Task-1 (Kirana Store Basics)

Create 'kirana_basics.ipynb' for a fictional kirana store whose name you invent. In one cell define store_name, item_name, price_inr, quantity_kg, is_organic and gst_percent in snake_case with a '#' comment above each line giving its unit, print one line per variable, then reassign item_name to a number and print it again to see dynamic typing.

```python
# Unit: text; fictional store name
store_name = "Sunrise Kirana"

# Unit: text; item name
item_name = "rice"

# Unit: rupees per kilogram (INR/kg)
price_inr = 46

# Unit: kilograms (kg)
quantity_kg = 12.5

# Unit: True or False
is_organic = True

# Unit: percent (%)
gst_percent = 5

print("Store name:", store_name)
print("Item name:", item_name)
print("Price (INR/kg):", price_inr)
print("Quantity (kg):", quantity_kg)
print("Organic:", is_organic)
print("GST (%):", gst_percent)

# Unit: number; demonstrates changing item_name's value type
item_name = 501
print("item_name after reassignment:", item_name)
```

#Task-2

Add a cell that calculates basic_amount, gst_amount and final_amount for 12.5 kg of rice at Rs 46 per kg with 5 percent GST, using your variables, and prints each amount rounded to 2 decimals with `round()`. Start the cell with a 3-line triple-quoted note stating what the cell does. Constraint: Do not use any `import`.

```python
"""
Calculate the basic amount for 12.5 kg of rice.
Calculate GST using the percentage from Task 1.
Print each amount rounded to two decimals.
"""

basic_amount = price_inr * quantity_kg
gst_amount = basic_amount * gst_percent / 100
final_amount = basic_amount + gst_amount

print("Basic amount (INR):", round(basic_amount, 2))
print("GST amount (INR):", round(gst_amount, 2))
print("Final amount (INR):", round(final_amount, 2))
```
